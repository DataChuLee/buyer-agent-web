#### Exa Search MCP Test

```python
SMITHERY_API_KEY = os.environ["SMITHERY_API_KEY"]
NAMESPACE = "stingray-C7LG"  # smithery namespace
CONNECTION_ID = "exa"  # smithery connectionId (해당 MCP ID)

client = MultiServerMCPClient(
    {
        "exa": {
            "transport": "streamable_http",
            "url": f"https://api.smithery.ai/connect/{NAMESPACE}/{CONNECTION_ID}/mcp",
            "headers": {
                "Authorization": f"Bearer {SMITHERY_API_KEY}",
                "Content-Type": "application/json",
            },
        }
    }

)
tools = await client.get_tools()
print([t.name for t in tools])
```

```python
exa_search_tool = next(t for t in tools if t.name == "web_search_exa")

result = await exa_search_tool.ainvoke(
    {
        "query": "10만원대 나이키 축구화 추천",
        "type": "auto",
        "numResults": 5,
    }
)
print(result)
```

#### Sequential Thinking MCP Test
- sse 방식은 jupyter 환경에서 실행됨
- 근데, stdio 방식은 jupyter 환경에서 error 발생
    - window 환경과 mcp stdio subprocess가 원인
    - 해결점은 우선 py 파일로 실행하는 것

```python
client = MultiServerMCPClient(
    {
        "sequential_thinking": {
            "transport": "stdio",
            "command": "npx.cmd",
            "args": ["-y",
                     "@smithery/cli@latest",
                     "run",
                     "@kiennd/reference-servers",
                     "--key",
                     f"{SMITHERY_API_KEY}"]
        }
    }

)
tools = await client.get_tools()
print([t.name for t in tools])
```
